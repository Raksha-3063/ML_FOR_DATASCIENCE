# Feature Encoding - Ordinal Encoding

## Introduction to Feature Encoding

Feature Encoding is a data preprocessing technique used to convert categorical data into numerical values so that machine learning algorithms can understand and process the information.

Most machine learning algorithms work with numerical data and cannot directly interpret text-based categories. Therefore, categorical features must be transformed into a numerical format before model training.

---

## What is Ordinal Data?

Ordinal data is a type of categorical data where the categories have a meaningful order or ranking.

In ordinal data, one category can be considered higher or lower than another based on a natural sequence.

### Examples

#### Education Level

- High School
- Bachelor's
- Master's
- PhD

#### Customer Satisfaction

- Poor
- Average
- Good
- Excellent

#### T-Shirt Size

- Small
- Medium
- Large

These categories follow a logical order, making them suitable for ordinal encoding.

---

## What is Ordinal Encoding?

Ordinal Encoding is a feature encoding technique used to convert ordinal categorical data into numerical values while preserving the natural order of the categories.

Each category is assigned a unique integer based on its ranking.

### Example

| Education Level | Encoded Value |
|----------------|---------------|
| High School | 1 |
| Bachelor's | 2 |
| Master's | 3 |
| PhD | 4 |

The numerical values represent the order of the categories rather than their magnitude.

---

## Why Use Ordinal Encoding?

- Converts ordered categorical data into numerical form.
- Preserves the ranking information present in the feature.
- Makes the data suitable for machine learning algorithms.
- Simple and efficient preprocessing technique.

# One Hot Encoding

## Introduction

One Hot Encoding is a feature encoding technique used to convert nominal categorical data into numerical format.

Unlike ordinal data, nominal categories do not have a natural order or ranking. Therefore, assigning numerical labels directly may introduce incorrect relationships between categories.

One Hot Encoding creates a separate binary column for each category.

### Example

| Gender |
|----------|
| Male |
| Female |

becomes

| Gender_Male |
|-------------|
| 1 |
| 0 |

Each encoded column contains either:

- 1 → Category is present
- 0 → Category is absent

This approach prevents machine learning algorithms from assuming any order between categories.

---

## Why Use One Hot Encoding?

- Suitable for nominal data.
- Prevents false ordering between categories.
- Easy to interpret.
- Widely used in machine learning preprocessing.

---

```python
job_counts = df['Job Title'].value_counts()
```

```python
threshold = 10
```

```python
common_jobs = job_counts[job_counts >= threshold].index

df['Job Title'] = df['Job Title'].apply(
    lambda x: x if x in common_jobs else 'Other'
)
```

```python
df['Job Title'].value_counts()
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(drop='first', sparse_output=False)

job_encoded = encoder.fit_transform(df[['Job Title']])
```

```python
job_encoded.shape
```

```python
job_encoded_df = pd.DataFrame(
    job_encoded,
    columns=encoder.get_feature_names_out(['Job Title'])
)

job_encoded_df.head()
```

```python
df = pd.concat([df, job_encoded_df], axis=1)
```

```python
df.drop('Job Title', axis=1, inplace=True)
```
