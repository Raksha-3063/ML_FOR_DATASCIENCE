## 📌 Overview

This project demonstrates **Feature Splitting** techniques using the famous Titanic dataset. Feature splitting is a feature engineering technique where a single feature is divided into multiple meaningful features to improve data representation and help machine learning models learn better patterns.

## 💡 Why Feature Splitting?

Many real-world datasets contain columns that hold multiple pieces of information in a single feature. Feature splitting separates these into individual, meaningful attributes, making the data easier for machine learning models to understand and often improving model performance.
