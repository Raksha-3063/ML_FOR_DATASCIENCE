# Missing Value Handling using KNN Imputer

## What is KNN Imputer?

**KNN Imputer (K-Nearest Neighbors Imputation)** is a technique used to fill missing values by looking at the **K nearest data points (neighbors)** that have similar feature values.

Instead of replacing missing values with the mean or median of the entire column, KNN Imputer estimates the missing value based on similar observations in the dataset.

It is available in the **scikit-learn** library under `sklearn.impute.KNNImputer`.

---

# How Does KNN Imputer Work?

The KNN Imputer follows these steps:

1. Identify the row containing the missing value.
2. Calculate the distance between this row and all other rows with complete values.
3. Find the **K nearest neighbors** based on the chosen distance metric (usually Euclidean distance).
4. Compute the average (or weighted average) of the neighbors' values for the missing feature.
5. Replace the missing value with the computed estimate.

---

# Example

Consider the following dataset:

| Age | Salary |
|-----|--------|
| 25 | 30000 |
| 30 | 40000 |
| NaN | 35000 |
| 35 | 45000 |

Suppose **K = 2**.

The algorithm finds the two most similar rows based on the available features.

Nearest neighbors:

- Age = 25
- Age = 30

Missing Age:

\[
\frac{25 + 30}{2} = 27.5
\]

The missing value is replaced with **27.5**.

---

# Why Use KNN Imputer?

Unlike mean or median imputation, KNN Imputer considers the similarity between observations, resulting in more realistic estimates.

It preserves relationships between features and often produces better results for machine learning models.

---

# Advantages

- Produces more accurate estimates than simple imputation methods.
- Preserves relationships between features.
- Works well for both small and medium-sized datasets.
- Reduces information loss.
- Simple to use with Scikit-learn.

---

# Disadvantages

- Computationally expensive for large datasets.
- Prediction time increases with dataset size.
- Sensitive to feature scaling.
- Not suitable for datasets with many missing values.
- Performance decreases in high-dimensional datasets.

---

# When to Use KNN Imputer?

Use KNN Imputer when:

- The dataset has a moderate number of missing values.
- Features are correlated.
- Preserving relationships between variables is important.
- Dataset size is relatively small or medium.

Avoid using it when:

- The dataset is extremely large.
- Features are not related.
- There are too many missing values.

---

# Feature Scaling

KNN Imputer uses distance calculations.

If one feature has much larger values than another, it can dominate the distance calculation.

Therefore, **feature scaling (Standardization or Normalization)** is recommended before applying KNN-based methods.

---

# Python Implementation

```python
from sklearn.impute import KNNImputer

# Create the imputer
imputer = KNNImputer(n_neighbors=5)

# Fit and transform the dataset
df_imputed = imputer.fit_transform(df)

# Convert back to DataFrame
import pandas as pd

df_imputed = pd.DataFrame(df_imputed, columns=df.columns)
```

---

# Important Parameters

| Parameter | Description |
|-----------|-------------|
| `n_neighbors` | Number of nearest neighbors to consider. |
| `weights` | Weight function (`uniform` or `distance`). |
| `metric` | Distance metric used to find neighbors (default: Euclidean). |

---


# Key Takeaways

- KNN Imputer estimates missing values using the **K nearest neighbors**.
- It uses **Euclidean distance** by default.
- It generally performs better than mean or median imputation because it considers relationships between observations.
- Feature scaling is recommended before using KNN-based methods.
- It is best suited for numerical datasets with moderate missing values.

````python
df_test = df.copy()
````

````python
original_age = df_test['Age'].copy()
````

````python
import numpy as np

np.random.seed(42)

mask = df_test['Age'].notnull()

random_index = df_test[mask].sample(30).index

df_test.loc[random_index, 'Age'] = np.nan
````

````python
from sklearn.impute import KNNImputer

numerical_df = df_test.select_dtypes(include=['number'])

imputer = KNNImputer(n_neighbors=5)

imputed = imputer.fit_transform(numerical_df)

imputed_df = pd.DataFrame(imputed, columns=numerical_df.columns)
````

````python
from sklearn.metrics import mean_absolute_error

true_values = original_age.loc[random_index]

predicted_values = imputed_df.loc[random_index, 'Age']

mae = mean_absolute_error(true_values, predicted_values)

print("Mean Absolute Error:", mae)
````

````python
from sklearn.metrics import root_mean_squared_error

rmse = root_mean_squared_error(true_values, predicted_values)

print("RMSE:", rmse)
````

````python
from sklearn.metrics import mean_squared_error
import numpy as np

rmse = np.sqrt(mean_squared_error(true_values, predicted_values))
print("RMSE:", rmse)
````

````python
from sklearn.impute import KNNImputer
from sklearn.metrics import r2_score
import pandas as pd

for k in [3, 5, 7, 9]:
    imputer = KNNImputer(n_neighbors=k)

    imputed = imputer.fit_transform(numerical_df)
    imputed_df = pd.DataFrame(imputed, columns=numerical_df.columns)

    predicted = imputed_df.loc[random_index, 'Age']

    r2 = r2_score(true_values, predicted)

    print(f"K = {k}, R² = {r2:.4f}")
````
