# Complete Case Analysis (CCA)

## What is Complete Case Analysis?

Complete Case Analysis (CCA), also known as **Listwise Deletion**, is a missing value handling technique where all rows containing one or more missing values are removed from the dataset.

Only observations with complete information are retained for further analysis and model training.

---

## When to Use Complete Case Analysis

CCA is suitable when:

* The percentage of missing values is very small (typically less than 5%).
* Missing values occur completely at random (MCAR).
* Removing rows does not significantly reduce the dataset size.
* The dataset is large enough to tolerate some data loss.

---

## Advantages

* Simple and easy to implement.
* No assumptions or predictions are required.
* Preserves the original values of the remaining observations.
* Fast preprocessing technique.

---

## Disadvantages

* Reduces the size of the dataset.
* Can remove a significant amount of useful information if many rows contain missing values.
* May introduce bias if the missing data is not completely random.
* Not recommended when the percentage of missing values is high.

---

## Python Implementation

```python
# Drop all rows containing missing values
df_cca = df.dropna()

# Display remaining missing values
df_cca.isnull().sum()
```

---

## Example

### Before Applying CCA

| Age | Salary | City      |
| --- | ------ | --------- |
| 25  | 30000  | Bangalore |
| NaN | 45000  | Mysore    |
| 30  | NaN    | Delhi     |
| 28  | 50000  | Mumbai    |

### After Applying CCA

| Age | Salary | City      |
| --- | ------ | --------- |
| 25  | 30000  | Bangalore |
| 28  | 50000  | Mumbai    |

Rows containing missing values are removed.

---

## When Not to Use CCA

Avoid CCA when:

* The dataset contains many missing values.
* The dataset is small.
* Missing values are not randomly distributed.
* Removing rows would significantly reduce model performance.

---

## Summary

Complete Case Analysis is one of the simplest techniques for handling missing data. It works by deleting all observations with missing values and is most effective when the proportion of missing data is very low and missingness occurs completely at random (MCAR). Although easy to implement, excessive row deletion can lead to information loss and biased models if used inappropriately.

# Numerical Data Imputation (Univariate Imputation)

## What is Univariate Imputation?

Univariate Imputation is a missing value handling technique in which missing values of a feature are replaced using information from the **same feature only**. It does not consider any relationship with other variables in the dataset.

This technique is commonly used for **numerical features** and is simple, fast, and easy to implement.

---

## Why Do We Need Numerical Imputation?

Machine Learning algorithms generally cannot handle missing numerical values directly. Imputing missing values helps to:

- Preserve valuable observations.
- Prevent model training errors.
- Improve model performance.
- Reduce information loss compared to deleting rows.

---

## Types of Univariate Numerical Imputation

### 1. Mean Imputation

Missing values are replaced with the **mean (average)** of the feature.

**Best suited for:**
- Normally distributed data.
- Data with no significant outliers.

**Advantages**
- Simple and fast.
- Preserves the overall mean.

**Disadvantages**
- Sensitive to outliers.
- Reduces variance.
- Can distort the original distribution.

---

### 2. Median Imputation

Missing values are replaced with the **median** of the feature.

**Best suited for:**
- Skewed distributions.
- Data containing outliers.

**Advantages**
- Robust to outliers.
- Preserves the central tendency of skewed data.

**Disadvantages**
- Reduces variance.
- Ignores relationships between features.

---

### 3. Arbitrary Value Imputation

Missing values are replaced with a predefined constant such as:

- -1
- 0
- 999
- 99999

The chosen value should lie outside the normal range of the feature.

**Advantages**
- Very simple to implement.
- Highlights missing values explicitly.
- Useful when missingness itself carries information.

**Disadvantages**
- Can introduce artificial outliers.
- May distort the feature distribution.

---

### 4. End of Distribution Imputation

Missing values are replaced with an extreme value located at the end of the feature's distribution.

A common approach is:

```
Mean + 3 × Standard Deviation
```

This ensures the imputed values lie outside the normal range of the data.

**Advantages**
- Preserves the information that values were originally missing.
- Useful for tree-based algorithms.

**Disadvantages**
- Introduces outliers.
- May negatively impact distance-based algorithms.

---

## Summary

| Technique | Best For | Sensitive to Outliers |
|-----------|----------|----------------------|
| Mean Imputation | Normally distributed data | Yes |
| Median Imputation | Skewed data | No |
| Arbitrary Value Imputation | When missing values have meaning | No |
| End of Distribution Imputation | Highlighting missing observations | No |

---

## Conclusion

Univariate Imputation is one of the most commonly used preprocessing techniques for handling missing numerical data. The choice of imputation method depends on the data distribution and the nature of the missing values. Mean and Median Imputation are suitable for replacing missing values with measures of central tendency, whereas Arbitrary Value and End of Distribution Imputation explicitly preserve the information that a value was originally missing.
