Regularization is a technique used to reduce **overfitting** in regression models.

In this notebook, we will study three regularized linear regression algorithms:

2. **Ridge Regression** → L2 Regularization
3. **ElasticNet Regression** → L1 + L2 Regularization

```text
Dataset

   ↓
Compare Models
```

All three models start from Linear Regression and add a **regularization penalty** to the loss function.

```text
Linear Regression
       +

       ↓
Lasso / Ridge / ElasticNet
```

* **Ridge → L2 → shrinks coefficients toward zero**

# Lasso Regression

## 1. What is Lasso Regression?

**Lasso** stands for **Least Absolute Shrinkage and Selection Operator**.

Lasso Regression is a regularized version of Linear Regression that uses **L1 Regularization**.

It is mainly used to:

* Reduce overfitting
* Control large model coefficients
* Perform feature selection
* Improve model generalization

### Key idea

> **Lasso = Linear Regression + L1 Regularization**

---

# 2. Why Do We Need Lasso?

In ordinary Linear Regression, the model tries to minimize the prediction error:

$$
MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

The model only cares about making predictions accurate.

However, if the model has many features, it may assign large coefficients to some features and become unnecessarily complex.

This can lead to **overfitting**.

Lasso solves this by adding a penalty for large coefficients.

So instead of minimizing only:

$$
MSE
$$

Lasso minimizes:

$$
MSE + \alpha\sum_{j=1}^{p}|\beta_j|
$$

where:

* $\alpha$ = regularization strength
* $\beta_j$ = model coefficient
* $|\beta_j|$ = absolute value of the coefficient
* $\sum|\beta_j|$ = **L1 penalty**

---

# 3. What is L1 Regularization?

L1 regularization adds the **absolute values of the coefficients** to the loss function.

$$
L1\ Penalty = \sum_{j=1}^{p}|\beta_j|
$$

For example, if the coefficients are:

```text
β1 = 5
β2 = -3
β3 = 2
```

then:

$$
|5| + |-3| + |2| = 10
$$

The model is therefore encouraged to keep coefficients small.

---

# 4. The Most Important Property of Lasso

The special property of Lasso is:

> **Lasso can make some coefficients exactly equal to 0.**

For example:

```text
Before Lasso:

Feature     Coefficient
X1             8.2
X2            -5.4
X3             0.3
X4             7.1
X5            -0.1
```

After Lasso:

```text
Feature     Coefficient
X1             6.9
X2            -4.7
X3             0.0
X4             5.8
X5             0.0
```

Here, `X3` and `X5` have coefficients of `0`.

Therefore, they do not contribute to the prediction.

This is called:

## Feature Selection

So remember:

> **Lasso can automatically remove less important features by making their coefficients zero.**

---

# 5. Lasso vs Linear Regression

### Linear Regression

```text
Prediction Error
       ↓
Minimize MSE
```

### Lasso Regression

```text
Prediction Error
       +
L1 Penalty
       ↓
Minimize MSE + α × Σ|β|
```

Therefore:

```text
Linear Regression
        ↓
No regularization

Lasso Regression
        ↓
L1 regularization
        ↓
Shrinks coefficients
        ↓
Some coefficients can become 0
        ↓
Feature selection
```

---

# 6. What is α (Alpha)?

`alpha` controls the **strength of regularization**.

In scikit-learn:

```python

```

Here:

```text
alpha = 1.0
```

### If α is small

```text

   ↓
Weak penalty
   ↓
Model behaves more like Linear Regression
```

### If α is large

```text
Large α
   ↓
Strong penalty
   ↓
Coefficients shrink more
   ↓
More coefficients may become 0
```

### Important

If `alpha` becomes too large:

```text
Too much regularization
        ↓
Model becomes too simple
        ↓
Underfitting
```

So `alpha` controls the **trade-off between fitting the data and keeping the model simple**.

---


# Ridge Regression

## 1. What is Ridge Regression?

**Ridge Regression** is a regularized version of Linear Regression that uses **L2 Regularization**.

It is mainly used to:

* Reduce overfitting
* Control large coefficients
* Improve model generalization
* Handle multicollinearity

### Key idea

> **Ridge = Linear Regression + L2 Regularization**

---

# 2. Why Do We Need Ridge Regression?

Ordinary Linear Regression tries to minimize only the prediction error.

When a model has many features or highly correlated features, some coefficients can become very large.

Large coefficients can make the model sensitive to small changes in the data and can increase the risk of overfitting.

Ridge addresses this by adding a penalty for large coefficients.

Therefore, Ridge tries to minimize:

$$
MSE + \alpha\sum_{j=1}^{p}\beta_j^2
$$

where:

* $\alpha$ = regularization strength
* $\beta_j$ = model coefficient
* $\beta_j^2$ = squared coefficient
* $\sum\beta_j^2$ = **L2 penalty**

---

# 3. What is L2 Regularization?

L2 regularization adds the **squared values of the coefficients** to the loss function.

$$
L2\ Penalty = \sum_{j=1}^{p}\beta_j^2
$$

For example, suppose:

```text
β1 = 5
β2 = -3
β3 = 2
```

Then:

$$
5^2 + (-3)^2 + 2^2
$$

$$
=25+9+4
$$

$$
=38
$$

The model is therefore encouraged to keep coefficients small.

---

# 4. The Most Important Property of Ridge

The most important characteristic of Ridge is:

> **Ridge shrinks coefficients toward zero but generally does not make them exactly zero.**

For example:

```text
Before Ridge:

Feature     Coefficient
X1             8.2
X2            -5.4
X3             3.1
X4             7.1
X5            -2.8
```

After Ridge:

```text
Feature     Coefficient
X1             6.5
X2            -4.2
X3             2.3
X4             5.7
X5            -1.9
```

The coefficients became smaller, but they are still present.

Compare this with Lasso:

```text
Lasso:

X1      6.9
X2     -4.7
X3      0
X4      5.8
X5      0
```

### Therefore:

> **Lasso can eliminate features. Ridge mainly shrinks their influence.**

---

# 5. Ridge vs Linear Regression

### Linear Regression

```text
Prediction Error
       ↓
Minimize MSE
```

### Ridge Regression

```text
Prediction Error
       +
L2 Penalty
       ↓
Minimize MSE + α × Σβ²
```

So:

```text
Linear Regression
        ↓
No regularization

Ridge Regression
        ↓
L2 regularization
        ↓
Penalizes large coefficients
        ↓
Shrinks coefficients
        ↓
Reduces overfitting
```

---

# 6. What is α (Alpha)?

`alpha` controls the strength of the L2 penalty.

In scikit-learn:

```python
Ridge(alpha=1.0)
```

Here:

```text
alpha = 1.0
```

### Small α

```text

   ↓

   ↓

   ↓
Model behaves more like Linear Regression
```

### Large α

```text
Large α
   ↓
Strong regularization
   ↓
More coefficient shrinkage
   ↓
Simpler model
```

If `alpha` becomes too large:

```text
Too much regularization
        ↓
Coefficients become too small
        ↓
Model may underfit
```

Therefore:

> **Alpha controls the trade-off between fitting the training data and keeping coefficients small.**

---

# 7. Ridge and Multicollinearity

One of the important applications of Ridge is dealing with **multicollinearity**.

### What is multicollinearity?

Multicollinearity occurs when two or more features are highly correlated.

For example:

```text
House Size  ────────┐
                    ├── Similar information
Number of Rooms ────┘
```

Ordinary Linear Regression may produce unstable or very large coefficients.

Ridge penalizes large coefficients and distributes the influence more smoothly across correlated features.

Therefore:

> **Ridge is particularly useful when predictors are strongly correlated.**

---

# 8. Ridge Does NOT Perform Feature Selection

This is an important difference between Ridge and Lasso.

### Lasso

```text
Coefficient
     ↓
Can become exactly 0
     ↓
Feature removed
```

### Ridge

```text
Coefficient
     ↓
Usually becomes smaller
     ↓

```

| Useful with multicollinearity  | Yes     | **Yes**    |   |           |

# ElasticNet Regression

## 1. What is ElasticNet Regression?

**ElasticNet Regression** is a regularized version of Linear Regression that combines both:

* **L1 Regularization → Lasso**
* **L2 Regularization → Ridge**

Therefore:

> **ElasticNet = Lasso + Ridge**

It is useful when we want the benefits of both L1 and L2 regularization.

---

# 2. Why Do We Need ElasticNet?

We have already seen:

### Lasso

Uses L1 regularization:

$$
MSE + \alpha\sum|\beta_j|
$$

Lasso can make some coefficients exactly zero.

Therefore, it can perform **feature selection**.

### Ridge

Uses L2 regularization:

$$
MSE + \alpha\sum\beta_j^2
$$

Ridge shrinks coefficients toward zero but generally keeps them non-zero.

It is particularly useful when features are correlated.

### The problem

Sometimes we want:

* Lasso's ability to perform feature selection
* Ridge's ability to handle correlated features

ElasticNet combines both.

---

# 3. ElasticNet Objective Function

ElasticNet adds both L1 and L2 penalties.

Conceptually:

$$
MSE
+
\alpha
\left[
\rho\sum|\beta_j|
+
(1-\rho)\sum\beta_j^2
\right]
$$

where:

* $\alpha$ controls the **overall regularization strength**
* $\rho$ controls the **balance between L1 and L2**

In scikit-learn, this balance parameter is called:

```python
l1_ratio
```

So remember:

```text
alpha     → How much regularization?
l1_ratio  → Which type of regularization?
```

---

# 4. Understanding `l1_ratio`

`l1_ratio` controls how much L1 and L2 regularization are used.

Its value is between:

$$
0 \leq l1\_ratio \leq 1
$$

### When `l1_ratio = 1`

The model behaves like Lasso:

```text
100% L1
0% L2
```

```text
ElasticNet
    ↓
L1 only
    ↓
Lasso-like behavior
```

### When `l1_ratio = 0`

The model becomes Ridge-like:

```text
0% L1
100% L2
```

```text
ElasticNet
    ↓
L2 only
    ↓
Ridge-like behavior
```

### When `l1_ratio = 0.5`

The model combines both:

```text
50% L1
50% L2
```

Therefore:

```text
l1_ratio = 0
       ↓
Ridge-like

l1_ratio = 0.5
       ↓
L1 + L2

l1_ratio = 1
       ↓
Lasso-like
```

---

# 5. `alpha` vs `l1_ratio`

This distinction is extremely important.

There are **two different controls** in ElasticNet.

### `alpha`

Controls:

> **How strong is the overall regularization?**

```text
Small alpha
   ↓
Weak regularization

Large alpha
   ↓
Strong regularization
```

### `l1_ratio`

Controls:

> **How much L1 vs L2 regularization is used?**

```text
l1_ratio = 1
       ↓
More L1

l1_ratio = 0
       ↓
More L2
```

### Memory trick

> **Alpha = Amount**
> **L1 ratio = Balance**

---

# 6. ElasticNet Intuition

Suppose we have:

```text
X1 ────────┐
           │
X2 ────────┤ → Target
           │
X3 ────────┤
           │
X4 ────────┘
```

ElasticNet tries to:

1. Fit the target well
2. Keep coefficients small
3. Potentially eliminate some features
4. Handle correlated features better than pure Lasso in some situations

Conceptually:

```text
Prediction Error
       +
      L1
       +
      L2
       ↓
ElasticNet
```

---

This is the most important comparison.

| ----------------- | -------------- | -----------------------: | ------------------- |
| Linear Regression | None           |                       No | Fit the data        |
| Lasso             | L1             |                  **Yes** | Feature selection   |
| Ridge             | L2             |               Usually No | Shrink coefficients |
| ElasticNet        | L1 + L2        |     **Yes, potentially** | Combination         |

### Memory

```text

  ↓
L1
  ↓
Remove some features


Ridge
  ↓
L2
  ↓
Reduce coefficient size


ElasticNet
  ↓
L1 + L2
  ↓
Remove + shrink
```

# 8. Why ElasticNet Can Be Useful with Correlated Features

Suppose we have:

```text
Feature A ────────┐
                  │
                  ├── Very similar information
Feature B ────────┘
```

These features are highly correlated.

Pure Lasso may select one feature and reduce another to zero.

Ridge tends to distribute the coefficients across correlated features.

ElasticNet combines these behaviors.

Therefore:

> ElasticNet can be useful when there are many features, some features are correlated, and we also want sparsity/feature selection.

# 🧠 Final Memory Trick

Remember these three words:

### LASSO → REMOVE

```text
L1
 ↓
Absolute value
 ↓
Some coefficients → 0
 ↓
Feature selection
```

### RIDGE → REDUCE

```text
L2
 ↓
Squared coefficients
 ↓
Coefficients shrink
 ↓
Usually remain non-zero
```

### ELASTICNET → COMBINE

```text
L1 + L2
 ↓
Remove + Reduce
 ↓
Feature selection + coefficient shrinkage
```

### One-line interview answer

> **Lasso uses L1 regularization and can perform feature selection by making coefficients zero, Ridge uses L2 regularization to shrink coefficients and is useful for multicollinearity, while ElasticNet combines both L1 and L2 regularization.**
