# Why Do We Handle Mixed Data?

## Introduction

In real-world datasets, it is common to find features that contain multiple types of information within a single column. These are known as **mixed data** or **mixed variables**. For example, a value such as **`C85`** contains two different pieces of information:

* **C** → Represents the cabin deck (categorical information).
* **85** → Represents the cabin number (numerical information).

Although humans can easily understand these components, machine learning algorithms treat the entire value (`C85`) as a single string and cannot recognize the hidden information within it.

## Why Is It Important?

If mixed values are left unchanged, the model considers each unique value as a completely different category. As a result:

* Useful patterns remain hidden.
* The model cannot distinguish between categorical and numerical information.
* Similar records may be treated as unrelated.
* Model performance and generalization may decrease.

By separating the mixed values into individual features, each feature represents a single type of information. This allows the model to learn meaningful relationships from the data.

## Why Should We Learn This?

Handling mixed data is an important feature engineering technique because most real-world datasets are not perfectly structured. Columns such as product codes, employee IDs, vehicle registration numbers, flight numbers, ticket IDs, and cabin numbers often combine letters and numbers.

Learning this technique helps us:

* Prepare raw data for machine learning.
* Extract hidden information from a dataset.
* Create meaningful features that improve model learning.
* Build preprocessing pipelines used in real-world machine learning projects.
* Develop practical data cleaning and feature engineering skills required in data science and machine learning roles.

## Conclusion

Handling mixed data is not simply about splitting a column into two parts. It is a feature engineering process that transforms complex, unstructured information into meaningful features that machine learning algorithms can understand and learn from effectively.


```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Cabin': [
        'C85', 'C123', 'B20', 'A10', 'C150',
        'B55', np.nan, 'D45', 'E12', 'F101',
        'A5', 'B89', 'C200', np.nan, 'E75',
        'D90', 'F34', 'A120', 'B15', 'C67'
    ],

    'Ticket': [
        'PC17599', 'A21171', '373450', 'PC17601', '113803',
        '347082', 'STON123', 'CA2144', 'LINE330', '113781',
        'PC19876', 'WEP563', 'A99881', '349209', 'SOTON221',
        'CA9987', '248706', 'PC45678', 'STON876', 'A14567'
    ],

    'Fare': [
        72.50, 85.20, 10.50, 15.00, 95.40,
        20.60, 50.00, 40.75, 18.90, 120.50,
        65.40, 33.25, 150.00, 12.75, 48.30,
        77.60, 29.50, 99.90, 25.80, 54.10
    ]
})

df
```

```python
df['Cabin_Letter'] = df['Cabin'].str.extract(r'([A-Za-z]+)')
```

```python
df['Cabin_Number'] = df['Cabin'].str.extract(r'(\d+)')
df['Cabin_Number'] = pd.to_numeric(df['Cabin_Number'])
```

```python
df['Ticket_Prefix'] = df['Ticket'].str.extract(r'([A-Za-z]+)')
```

```python
df['Ticket_Number'] = df['Ticket'].str.extract(r'(\d+)')
df['Ticket_Number'] = pd.to_numeric(df['Ticket_Number'])
```
